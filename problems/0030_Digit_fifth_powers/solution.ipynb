# Problem 30: Digit fifth powers


## Problem

Consider all integer combinations of $a^b$ for $2 ≤ a ≤ 5$ and $2 ≤ b ≤ 5$:

$2^2=4$, $2^3=8$, $2^4=16$, $2^5=32$
$3^2=9$, $3^3=27$, $3^4=81$, $3^5=243$
$4^2=16$, $4^3=64$, $4^4=256$, $4^5=1024$
$5^2=25$, $5^3=125$, $5^4=625$, $5^5=3125$


If they are then placed in numerical order, with any repeats removed, we get the following sequence of 15 distinct terms:

$$4, 8, 9, 16, 25, 27, 32, 64, 81, 125, 243, 256, 625, 1024, 3125$$

How many distinct terms are in the sequence generated by $a^b$ for $2 ≤ a ≤ 100$ and $2 ≤ b ≤ 100$?

## Implementation

We establish a hard limit in terms of number of digits we can reach by summing $9^power$. For example, the max number since $9^4 = 6561$, we cannot write any 6-digit number as the sum of the fourth powers of the digits (the highest number we can reach is ($6*9^4 = 39366$, a 5-digit number).

```python
# Solution
def digit_fifth_powers(power: int) -> int:
    import math

    max_digits = int(math.log10(9**power))+int(math.log10(power))+2
    return sum(i for i in range(max_digits*(9**power)) if i == sum(int(c)**power for c in str(i)))-1
# END solution

from IPython.display import  Markdown

POWER = 5
solution = digit_fifth_powers(power=POWER)

Markdown(f"""
## Solution

Sum of numbers that can be written as the sum of {POWER}-th powers: {solution}
""")
```
