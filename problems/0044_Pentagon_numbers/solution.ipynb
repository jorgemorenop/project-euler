# Problem 44: Pentagon numbers

## Problem

Pentagonal numbers are generated by the formula, $P_n=\frac{n(3n−1)}{2}$. The first ten pentagonal numbers are:

$$1, 5, 12, 22, 35, 51, 70, 92, 117, 145, ...$$

It can be seen that $P_4 + P_7 = 22 + 70 = 92 = P_8$. However, their difference, $70 − 22 = 48$, is not pentagonal.

Find the pair of pentagonal numbers, $P_j$ and $P_k$, for which their sum and difference are pentagonal and $D = |P_k − P_j|$ is minimised; what is the value of $D$?

## Implementation

```python
# Solution
def pentagon_numbers() -> int:
    import math

    def is_pentagon(number: int):
        # This is the quadratic solution for n of n(3n-1)/2 = s
        return (1 + math.sqrt(1 + 24 * number)) % 6 == 0 # We don't consider the (-) bc n must be positive

    def p(n: int):
        return n*(3*n-1)

    min_d = float("inf")
    p_n = [1, 5]
    while p_n[-1]-p_n[-2] < min_d:
        for j in range(2, len(p_n)+1):
            d = p_n[-1]-p_n[-j]
            if d >= min_d:
                break
            if is_pentagon(d) and is_pentagon(p_n[-1]+p_n[-j]):
                print(f"k={p_n[-1]}, j={p_n[-j]}, d={d}, s={p_n[-1]+p_n[-j]}")
                min_d = d
        p_n.append(p_n[-1]+3*len(p_n)+1)
    # print(p_n)
    return min_d
# END solution


from IPython.display import  Markdown

solution = pentagon_numbers()

Markdown(f"""
## Solution

D = {solution}
""")
```
